# 02 QA Dataset EDA

This notebook presents a publication-ready exploratory data analysis for the final QA dataset release. The underlying figures and tables are generated by a reproducible script rather than being drawn manually inside the notebook.

```python
from __future__ import annotations

import json
import subprocess
import sys
from pathlib import Path

import pandas as pd
from IPython.display import Image, display

ROOT = Path.cwd().resolve()
if not (ROOT / "data").exists():
    ROOT = ROOT.parent.resolve()

sys.path.append(str(ROOT))

EDA_DIR = ROOT / "eda" / "figures" / "02_qa_dataset_eda"
REPORT_PATH = EDA_DIR / "eda_report.json"

if not REPORT_PATH.exists():
    subprocess.run([sys.executable, str(ROOT / "scripts" / "eda" / "build_qa_dataset_eda.py")], check=True)

report = json.loads(REPORT_PATH.read_text(encoding="utf-8"))
report
```

## Analytical Scope

- **Overview charts** are computed from the final three-way release dataset.
- **Quality diagnostic charts** are computed from the full Gemini-annotated pool before weak-quality filtering, so that pipeline weaknesses remain visible rather than being hidden by the final release policy.
- This split is intentional and should be stated explicitly in the report.

## Suggested Reporting Narrative

1. Start from dataset scale and domain coverage to show that the final release is broad rather than narrow.
2. Show that the dataset is structurally non-trivial because reasoning complexity is not uniform across the release.
3. Use the quality diagnostics to argue that multi-sentence inference is materially harder for LLM generation than extraction.
4. Use the domain quality chart to identify pipeline limitations transparently.
5. Close the section by connecting these observations to the need for human verification and inter-annotator agreement analysis.
